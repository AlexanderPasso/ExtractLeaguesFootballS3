```python
import pandas as pd
import time 
import random
import os
from datetime import datetime
import uuid
```

```python
url = ['https://www.espn.com.co/futbol/posiciones/_/liga/col.1',
       'https://www.espn.com.co/futbol/posiciones/_/liga/eng.1',
       'https://www.espn.com.co/futbol/posiciones/_/liga/esp.1',
       'https://www.espn.com.co/futbol/posiciones/_/liga/ger.1',
       'https://www.espn.com.co/futbol/posiciones/_/liga/fra.1',
       'https://www.espn.com.co/futbol/posiciones/_/liga/ita.1'
]
ligas = ['COLOMBIA','INGLATERRA','ESPAÑA','ALEMANIA','FRANCIA','ITALIA']

df_ligas = {
    'LIGA': ligas,
    'URL':url
}

df_ligas = pd.DataFrame(df_ligas)
df_ligas.to_csv("../data/df_ligas.csv", index=False)
```

```python
#df_ligas['URL'][2]
df = pd.read_html(df_ligas['URL'][0])
#len(df)
#df[0] #Dataframe nombre de equipos
#df[1]  #Dataframe de puntajes e informacion numerica

df = pd.concat([df[0],df[1]], ignore_index=True, axis=1)

#Renombrar columnas
df = df.rename(columns={0:'EQUIPO', 1:'J', 2:'G',3:'E', 4:'P', 5:'GF',6:'GC',7:'DIF',8:'PTS'})
df['EQUIPO'] = df['EQUIPO'].apply(lambda x: x[5:] if x[:2].isnumeric() ==True else x[4:])

#Agregar nombre de liga
df['LIGA'] = df_ligas['LIGA'][0]
df
```

```python
#Generar los Id's unicos para cada equipo

import secrets
import string

# Definir letras y números (62 caracteres posibles por posición)
caracteres = string.ascii_letters + string.digits

# Generar un ID seguro de exactamente 30 caracteres
id_de_30 = ''.join(secrets.choice(caracteres) for _ in range(8))

print(id_de_30)
# Resultado de ejemplo: kF7sN2mP9xL1qW5vT8zB4jY6cR3eA0
print(len(str(id_de_30)))


# Generar un ID único de 8 caracteres para cada fila
df['ID'] = df.apply(lambda row: ''.join(secrets.choice(caracteres) for _ in range(8)), axis=1)
df[['EQUIPO','ID']].to_csv("../data/team_table.csv", index=False)
```
